```python
from math import ceil
from pathlib import Path
import pickle
import matplotlib.pyplot as plt

logs_dir = Path(r"C:\Users\nly99857\OneDrive - Philips\SW Projects\QSeaBattle\WIP\logs")
log_files = sorted(logs_dir.glob("train_run_*_*.pkl"))

if not log_files:
    raise FileNotFoundError(f"No training logs found in {logs_dir}")

plot_data = []
skipped_logs = []

for log_file in log_files:
    try:
        with log_file.open("rb") as file_handle:
            data = pickle.load(file_handle)
    except Exception as exc:
        skipped_logs.append(
            f"{log_file.name}: failed to load pickle ({type(exc).__name__}: {exc})"
        )
        continue

    if isinstance(data, dict) and "epochs" in data:
        records = data["epochs"]
    elif isinstance(data, list):
        records = data
    else:
        skipped_logs.append(
            f"{log_file.name}: unsupported root type {type(data).__name__}"
        )
        continue

    if not records:
        skipped_logs.append(f"{log_file.name}: no epoch records found")
        continue

    try:
        epochs = [entry["epoch"] for entry in records]
        shoot_acc = [entry["shoot_acc"] for entry in records]
    except (TypeError, KeyError) as exc:
        skipped_logs.append(f"{log_file.name}: missing expected fields ({exc})")
        continue

    plot_data.append((log_file.name, epochs, shoot_acc))

if not plot_data:
    raise ValueError("No logs with plottable epoch and shoot_acc data were found.")

print("Files being plotted:")
for file_name, _, _ in plot_data:
    print(f"- {file_name}")

plots_per_figure = 12
ncols = 3

for start_index in range(0, len(plot_data), plots_per_figure):
    batch = plot_data[start_index : start_index + plots_per_figure]
    nrows = ceil(len(batch) / ncols)
    fig, axes = plt.subplots(
        nrows=nrows,
        ncols=ncols,
        figsize=(5.5 * ncols, 3.4 * nrows),
        squeeze=False,
    )
    flat_axes = axes.flatten()

    for axis, (file_name, epochs, shoot_acc) in zip(flat_axes, batch):
        axis.plot(epochs, shoot_acc, marker="o", markersize=2.2, linewidth=1)
        axis.set_title(file_name, fontsize=9)
        axis.set_xlabel("Epoch")
        axis.set_ylabel("Shoot Accuracy")
        axis.grid(True, alpha=0.3)

    for axis in flat_axes[len(batch):]:
        axis.remove()

    figure_number = start_index // plots_per_figure + 1
    total_figures = ceil(len(plot_data) / plots_per_figure)
    fig.suptitle(
        f"Shoot Accuracy vs Epoch (figure {figure_number} of {total_figures})",
        fontsize=14,
    )
    fig.tight_layout(rect=(0, 0, 1, 0.97))
    plt.show()

print(f"Plotted {len(plot_data)} log files from {logs_dir}")
if skipped_logs:
    print("Skipped logs:")
    for message in skipped_logs:
        print(f"- {message}")
else:
    print("Skipped logs: none")
```
