```python
from pathlib import Path
import pickle

import matplotlib.pyplot as plt

log_path = Path(r"C:\Users\nly99857\OneDrive - Philips\SW Projects\QSeaBattle\WIP\logs\train_run_20260308_105339.pkl")

with log_path.open("rb") as f:
    run_log = pickle.load(f)

epochs_data = run_log["epochs"]
epoch_idx = [row["epoch"] for row in epochs_data]
shoot_acc = [row["shoot_acc"] for row in epochs_data]

print(f"Loaded {len(epochs_data)} epochs from: {log_path}")
print(f"Shoot accuracy range: {min(shoot_acc):.4f} to {max(shoot_acc):.4f}")
```

```python
plt.figure(figsize=(10, 5))
plt.plot(epoch_idx, shoot_acc, linewidth=1.8, color="tab:blue", label="Shoot Accuracy")
plt.plot(epoch_idx, [0.9703] * len(epoch_idx), linestyle="--", color="tab:orange", label="Target Accuracy (0.90)")
plt.title("Shoot Accuracy Over Epochs")
plt.xlabel("Epoch")
plt.ylabel("Shoot Accuracy")
plt.ylim(0.85, 1.0)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
